### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def count_k(n):
    count=0
    for k in range(1,n):
        if k>(n-1)/2:
            count+=1
    return count
n = int(float(input()))
print(count_k(n))
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math
def compute_area(a, b, c):
    if a+b>c and b+c>a and c+a>b:
        p = (a+b+c)/2
        return math.sqrt(p*(p-a)*(p-b)*(p-c))
    else:
        raise ValueError('Sides cannot form a triangle')
a = float(input())
b = float(input())
c = float(input())
print(compute_area(a, b,c))
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
def convert(x):
    if len(x)==1 and x.isalpha():
        return x.lower()
    else:
        raise ValueError('Is not a single character')
x = input()
print(convert(x))
```

### Problem 10
#### Write a function that, given a lowercase letter, returns a string containing that letter repeated 10 times.

```python
def repeat(x):
    if x.islower():
        return x*10
    else: return 'Not a lowercase letter'
x = input()
print(repeat(x))
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
def triangle(n):
    line = []
    for i in range(1,n+1):
        line.append('*'*i)
    return'\n'.join(line)
n = int(float(input()))
print(triangle(n))
```
