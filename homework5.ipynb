```python
def armstrong(x: int):
    n = len(str(x))
    total = 0
    for i in str(x):
        total += int(i)**n
    if total == x:
        return True
    return False
```

```python
import math
def goc(A:tuple, B:tuple, C:tuple):
    xa = A[0]
    ya = A[1]
    xb = B[0]
    yb = B[1]
    xc = C[0]
    yc = C[1]
    ba = (xa-xb, ya-yb)
    bc = (xc-xb, yc-yb)
    x1 = ba[0]
    y1 = ba[1]
    x2 = bc[0]
    y2 = bc[1]
    ba = math.sqrt(x1**2+y1**2)
    bc = math.sqrt(x2**2+y2**2)
    if ba == 0 or bc == 0:
        return 'Invalid'
    else:
        cos_ABC= (x1*x2+y1*y2)/(ba*bc)
        rad = math.acos(cos_ABC)
        degree = math.degrees(rad)
        return degree
```

```python
def max_min(a,b,c):
    max = a
    if b >= max:
        max = b
    if c >= max:
        max = c
    min = a
    if b <= min:
        min = b
    if c <= min:
        min = c
    return [max, min]
```

```python
def check(n: int):
    lst = list(str(n))
    s = set(lst)
    if len(lst) == len(s):
        return True
    return False
def sum_1_to_n(n):
    if n<=0:
        return 0
    else:
        sum = 0
        for i in range(1,n+1):
            if check(i):
                sum+=i
        return sum
print(sum_1_to_n(100))
```
