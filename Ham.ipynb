```python
def snt(n):
    if n < 2:
        return False
    else:
        for i in range(1,n):
            if n%i==0:
                return False
    return True
```

```python
def check_hh(a):
    lst = []
    for i in range(1,a):
        if a%i==0:
            lst.append(i)
    sum = 0
    for i in lst:
        sum+=i
    if sum==a:
        return True
    else:
        return False
```

```python
def divide_with_remainder(dividend: float, divisor:float):
    if divisor == 0:
        return 'Invalid'
    else:
        q = dividend//divisor
        r = dividend%divisor
        return f'({q}, {r})'
```

```python
def is_right_triangle(a, b, c):
    if a+b>c and b+c>a and c+a>b:
        if a**2+b**2==c**2 or b**2+c**2==a**2 or c**2+a**2==b**2:
            return True
    return False
```
