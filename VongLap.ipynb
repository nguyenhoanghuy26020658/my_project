```python
def check(n):
    s = str(n)
    if len(set(s)) == len(s):
        return True
    return False
def check_cp(n):
    if n<0:
        return False
    else:
        root = int(n**0.5)
        if root*root == n:
            return True
        else:
            return False
n = int(input())
if n>0:
    lst = []
    for i in range(1,n+1):
        if check_cp(i)==True:
            lst.append(i)
    lst2 = []
    for j in lst:
        if check(j)==True:
            lst2.append(str(j))
    print(' '.join(lst2))
else:
    print('n khong la so nguyen duong')
```

```python
def collatz(x):
    count = 1
    while x!=1:
        if x%2==0:
            x=x//2
            count+=1
        else:
            x=3*x+1
            count+=1
    return count
n = int(input())
if n>0:
    lst = []
    lst2 = []
    for i in range(1,n+1):
        lst.append(collatz(i))
    for i in range(1,n+1):
        if collatz(i) == max(lst):
            lst2.append(i)
    print(f'So co do dai collatz lon nhat: {min(lst2)}')
    print(f'Do dai collatz = {max(lst)}')
else:
    print('n khong hop le')
```

```python
def check(a,b):
    lst = []
    for i in range(1,a):
        if a%i==0:
            lst.append(i)
    total = 0
    for i in lst:
        total+=i
    if total==b:
        return True
    else:
        return False

a, b = map(float, input().split())
if check(int(a),int(b))==True and check(int(b),int(a))==True:
    print('True')
else:
    print('False')
```

```python
def uoc(a):
    lst = []
    for i in range(1,int(a)+1):
        if int(a)%i==0:
            lst.append(i)
    return lst
m, n = map(float, input().split())
new_lst = []
for i in uoc(m):
    if i in uoc(n):
        new_lst.append(i)
print(max(new_lst))
```
