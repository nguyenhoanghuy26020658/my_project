W4A18

```python
def tim_uoc(n):
    uoc = []
    for i in range(1,n+1):
        if n%i==0:
            uoc.append(i)
    return uoc
a, b = map(int, input().split())
for i in tim_uoc(a):
    if i in tim_uoc(b):
        print(i)
```

W4A20

```python
def check(n):
    if n==1:
        return True
    elif n>1:
        while True:
            if n%2==0:
                n=n/2
                continue
            elif n==1:
                return True
                break
            else:
                return False
                break
    else:
        return False
n = int(input())
if check(n):
    print('Yes')
else:
    print('No')
```

W4A29

```python
n = input().strip()
lst = []
for i in n:
    if 65<=ord(i)<=90:
        i = i.lower()
        lst.append(i)
    elif 97<=ord(i)<=122:
        i = i.upper()
        lst.append(i)
    elif i == ' ':
        lst.append(i)
print(''.join(lst))
    
```
