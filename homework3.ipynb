Bài 8: Nhập vào số nguyên dương a và b, in toàn bộ ước dương chung của a và b

```python
def uoc(a):
    lst = []
    for i in range(1,a+1):
            if a%i==0:
                lst.append(i)
    return lst
a, b = map(float, input().split())
if a>0 and b>0:
    new_lst = []
    for i in uoc(int(a)):
        if i in uoc(int(b)):
            new_lst.append(str(i))
    print(' '.join(new_lst))
else:
    print('a, b khong hop le')

      
      
```

Bài 9: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay
không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25…. 

```python
def check_cp(a):
    new_a = int(a**0.5)
    if new_a**2 == a:
        return True
    else:
        return False
a = int(input())
if a>0:
    if check_cp(a) == True:
        print('La so chinh phuong')
    else:
        print('Khong la so chinh phuong')
else:
    print('a khong hop le')
```

Bài 10: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 

```python
def check_hh(a):
    lst = []
    for i in range(1,a):
        if a%i==0:
            lst.append(i)
    sum = 0
    for i in lst:
        sum+=i
    if sum==a:
        return True
    else:
        return False
a = int(input())
if a>0:
    if check_hh(a) == True:
        print('La so hoan hao')
    else:
        print('Khong la so hoan hao')
else:
    print('a khong hop le')
```

Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k)
lớn nhất nhưng nhỏ hơn n

```python
n = int(input())
lst = []
for k in range(1,n):
    Sk = 0
    for i in range(1,k+1):
        Sk+=i
    if Sk<n:
        lst.append(k)
print(f'k = {max(lst)}')
```

Bài 13: Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho
1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A

```python
a = int(input())
n = 0
sum = 0
while sum<=a:
    sum = 0
    n+=1
    for i in range(1,n+1):
        sum+= 1/i
print(n)
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng 

```python
lst = []
while True:
    try:
        n = int(input())
        lst.append(n)
        if n == -1:
            break
        else:
            continue
    finally: pass
print(max(lst))
print(min(lst))
```

Bài 18: Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 3^k hay
không 

```python
def check(n):
    if n == 1:
        return True
    elif n>1:
        while True:
            try:
                if n%3==0:
                    n = n/3
                    continue
                elif n==1:
                    return True
                    break
                else: 
                    return False
                    break
            finally: pass
    else:
        return False
n = int(input())
if check(n)==True:
    print('Yes')
else:
    print('No')
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 


```python
a = int(input())
if a>0:
    lst = [1,1]
    for i in range(a):
        lst.append(lst[-1]+lst[-2])
    lst2 = []
    for j in lst:
        if j<=a:
            lst2.append(j)
    print(max(lst2))
else:
    print('A khong hop le')
```

Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố. 

```python
def snt(a):
    if a<2:
        return False
    for i in range(2,a):
        if a%i==0:
            return False
    return True
n = input()
if int(n)>0:
    lst = list(n)[::1]
    total = 0
    for i in lst:
        if snt(int(i))==True:
            total+=1
    print(f'So {n} co {total} chu so la so nguyen to')
else:
    print('n khong hop le')
```
