Bài 2: Một khu đất hình chữ nhật có chiều rộng a, chiều dài b. Người ra xây một khu
vui chơi hình tròn tại vị trí trung tâm khu đất như hình vẽ. Nhập vào a, b. Hãy tính phần diện tích còn để trồng cây xung quanh là bao nhiêu. Biết pi = 3.14, lấy kết quả đến 2 chữ số phần thập phân. 

```python
def tree_area(a, b):
    if a>0 and b>0:
        return f'Dien tich trong cay = {round((a*b)-(3.14*((a**2)/4)),2)}'
    else:
        return 'a, b khong hop le'
a = float(input('Dien chieu rong: '))
b = float(input('Dien chieu dai: '))
print(tree_area(a, b))
```

Bài 3: Viết chương trình nhập vào một “kí tự” c, nếu nó là chữ hoa thì in ra chữ thường tương ứng và ngược lại nếu nó là chữ thường thì in ra chữ hoa tương ứng của nó. (xem trước cấu trúc lựa chọn: if….else, ma) 


```python
def reverse(c):
    if c.isupper():
        return c.lower()
    if c.islower():
        return c.upper()
    else: 
        raise ValueError('c ko phai la mot ki tu')
c = input('Nhap mot ki tu: ')
print(reverse(c))
```

Bài 4: Viết chương trình nhập vào một “kí tự”, Kiểm tra xem c có phải là kí tự chữ cái không (hoa hoặc thường). Nếu đúng thì in ra “{c} là kí tu alphabet”, nếu sai in ra “{c} không phải là kí tự alphabet. 

```python
def check(c):
    if c.isalpha():
        return f'{c} la ki tu alphabet'
    else:
        return f'{c} khong phai la ki tu alphabet'
c = input('Nhap mot ki tu: ')
print(check(c))
```

Bài 5: Nhập vào chữ cái hoa A, in ra chữ cái thường liền trước chữ cái a thường tương ứng của A. Xét TH khi c = ‘A’ là đặc biệt. 

```python
c = input('Nhap chu cai hoa: ')
if c.isupper and c != 'A':
    c = ord(c.lower())-1
    print(f'Chu cai can tim: {chr(c)}')
elif c == 'A':
    print('Chu cai can tim: z')
else:
    print('Chu cai nhap vao khong phai chu cai hoa')
```

Bài 6. Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó không phải 3 cạnh của tam giác thì in ra “Khong phai 3 canh cua tam giac”. 


```python
import math
def check_triangle(a, b, c):
    if a+b>c and b+c>a and c+a>b:
        p = (a+b+c)/2
        return f'Dien tich tam giac = {math.sqrt(p*(p-a)*(p-b)*(p-c))}'
    else:
        return 'Khong phai 3 canh cua tam giac'
a = float(input())
b = float(input())
c = float(input())
print(check_triangle(a, b, c))
```

```python
name = input('Nhap ten chu ho: ')
a = float(input('Nhap chi so thang truoc: '))
b = float(input('Nhap chi so thang nay: '))
def calculate(a, b):
    c = b-a
    tien_dien = 0
    if (a<0 and b<0) or c<0:
        return 'Chi so khong hop le'
    if c>=0 and c<=50:
        tien_dien += c*1984
    if c>=51 and c<=100:
        tien_dien += 50*1984+(c-50)*2050
    if c>=101 and c<=200:
        tien_dien += 50*1984+50*2050+(c-100)*2380
    if c>=201 and c<=300:
        tien_dien += 50*1984+50*2050+100*2380+(c-200)*2998
    if c>=301 and c<=400:
        tien_dien += 50*1984+50*2050+100*2380+100*2998+(c-300)*3350
    if c>=401:
        tien_dien += 50*1984+50*2050+100*2380+100*2998+100*3350+(c-400)*3460
    return((tien_dien)*1.08)
print(f'Ho va ten: {name}')
print(f'Tien phai tra la: {round(calculate(a, b))}')
    
```
